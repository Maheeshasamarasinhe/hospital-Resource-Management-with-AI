```python
import keras_tuner as kt
from tensorflow import keras
from tensorflow.keras import layers, regularizers, callbacks
from sklearn.preprocessing import MinMaxScaler
import tensorflow as tf

# ===== DATA PREPARATION =====
seq_cols = target_cols + ['Rainfall', 'Avg_Temperature', 'Humidity']  # 10 Features
static_cols = ['Month_Sin', 'Month_Cos', 'Festive_Season', 'Public_Holidays', 'Public_Awareness_Level']  # 5 Features

# Scaling
scaler_seq = MinMaxScaler()
scaler_static = MinMaxScaler()
scaler_target = MinMaxScaler()

scaled_seq = scaler_seq.fit_transform(df[seq_cols])
scaled_static = scaler_static.fit_transform(df[static_cols])
scaled_target = scaler_target.fit_transform(df[target_cols])

# Sliding Window
def create_data(seq_in, static_in, target_in, window=12):
    X_s, X_st, y = [], [], []
    for i in range(len(seq_in) - window):
        X_s.append(seq_in[i:i+window])
        X_st.append(static_in[i+window])
        y.append(target_in[i+window])
    return np.array(X_s), np.array(X_st), np.array(y)

X_seq, X_static, y = create_data(scaled_seq, scaled_static, scaled_target)
split = int(len(X_seq) * 0.8)
X_seq_train, X_seq_test = X_seq[:split], X_seq[split:]
X_static_train, X_static_test = X_static[:split], X_static[split:]
y_train, y_test = y[:split], y[split:]

# ===== CUSTOM ATTENTION LAYER =====
class AttentionLayer(layers.Layer):
    """Attention mechanism to focus on important time steps"""
    def __init__(self, **kwargs):
        super(AttentionLayer, self).__init__(**kwargs)
    
    def build(self, input_shape):
        self.W = self.add_weight(name='attention_weight',
                                  shape=(input_shape[-1], input_shape[-1]),
                                  initializer='glorot_uniform',
                                  trainable=True)
        self.b = self.add_weight(name='attention_bias',
                                  shape=(input_shape[-1],),
                                  initializer='zeros',
                                  trainable=True)
        self.u = self.add_weight(name='attention_context',
                                  shape=(input_shape[-1], 1),
                                  initializer='glorot_uniform',
                                  trainable=True)
        super(AttentionLayer, self).build(input_shape)
    
    def call(self, inputs):
        # Attention scores
        score = tf.nn.tanh(tf.tensordot(inputs, self.W, axes=1) + self.b)
        attention_weights = tf.nn.softmax(tf.tensordot(score, self.u, axes=1), axis=1)
        # Weighted sum
        context_vector = tf.reduce_sum(inputs * attention_weights, axis=1)
        return context_vector

# ===== PROFESSIONAL HYBRID MODEL =====
def build_professional_model(hp):
    # Hyperparameters
    lstm_units_1 = hp.Int('lstm_units_1', min_value=64, max_value=256, step=32)
    lstm_units_2 = hp.Int('lstm_units_2', min_value=32, max_value=128, step=32)
    dense_units_static = hp.Int('dense_static', min_value=32, max_value=128, step=32)
    dense_units_combined = hp.Int('dense_combined', min_value=64, max_value=256, step=32)
    dropout_rate = hp.Float('dropout', min_value=0.1, max_value=0.5, step=0.1)
    l2_reg = hp.Float('l2_reg', min_value=1e-5, max_value=1e-3, sampling='log')
    learning_rate = hp.Choice('learning_rate', values=[1e-2, 5e-3, 1e-3, 5e-4])
    use_bidirectional = hp.Boolean('use_bidirectional', default=True)
    
    # ===== SEQUENTIAL INPUT BRANCH (LSTM) =====
    input_seq = keras.Input(shape=(12, len(seq_cols)), name='sequential_input')
    
    # First LSTM Layer (Bidirectional optional)
    if use_bidirectional:
        x_seq = layers.Bidirectional(
            layers.LSTM(lstm_units_1, return_sequences=True, 
                       kernel_regularizer=regularizers.l2(l2_reg))
        )(input_seq)
    else:
        x_seq = layers.LSTM(lstm_units_1, return_sequences=True,
                           kernel_regularizer=regularizers.l2(l2_reg))(input_seq)
    
    x_seq = layers.BatchNormalization()(x_seq)
    x_seq = layers.Dropout(dropout_rate)(x_seq)
    
    # Second LSTM Layer
    if use_bidirectional:
        x_seq = layers.Bidirectional(
            layers.LSTM(lstm_units_2, return_sequences=True,
                       kernel_regularizer=regularizers.l2(l2_reg))
        )(x_seq)
    else:
        x_seq = layers.LSTM(lstm_units_2, return_sequences=True,
                           kernel_regularizer=regularizers.l2(l2_reg))(x_seq)
    
    x_seq = layers.BatchNormalization()(x_seq)
    
    # Attention Layer
    x_seq = AttentionLayer()(x_seq)
    x_seq = layers.Dropout(dropout_rate)(x_seq)
    
    # ===== STATIC INPUT BRANCH (Dense Network) =====
    input_static = keras.Input(shape=(len(static_cols),), name='static_input')
    
    x_static = layers.Dense(dense_units_static, activation='relu',
                           kernel_regularizer=regularizers.l2(l2_reg))(input_static)
    x_static = layers.BatchNormalization()(x_static)
    x_static = layers.Dropout(dropout_rate * 0.5)(x_static)
    
    x_static = layers.Dense(dense_units_static // 2, activation='relu',
                           kernel_regularizer=regularizers.l2(l2_reg))(x_static)
    x_static = layers.BatchNormalization()(x_static)
    
    # ===== CONCATENATION & FUSION LAYERS =====
    combined = layers.concatenate([x_seq, x_static], name='fusion_layer')
    
    # Deep Fusion Network
    x = layers.Dense(dense_units_combined, activation='relu',
                    kernel_regularizer=regularizers.l2(l2_reg))(combined)
    x = layers.BatchNormalization()(x)
    x = layers.Dropout(dropout_rate)(x)
    
    x = layers.Dense(dense_units_combined // 2, activation='relu',
                    kernel_regularizer=regularizers.l2(l2_reg))(x)
    x = layers.BatchNormalization()(x)
    x = layers.Dropout(dropout_rate * 0.5)(x)
    
    x = layers.Dense(32, activation='relu',
                    kernel_regularizer=regularizers.l2(l2_reg))(x)
    
    # ===== OUTPUT LAYER =====
    output = layers.Dense(len(target_cols), activation='linear', name='predictions')(x)
    
    # Build Model
    model = keras.Model(inputs=[input_seq, input_static], outputs=output)
    
    # Compile with Huber Loss (robust to outliers)
    model.compile(
        optimizer=keras.optimizers.Adam(learning_rate=learning_rate),
        loss='huber',
        metrics=['mae', 'mse']
    )
    
    return model

# ===== CALLBACKS =====
early_stopping = callbacks.EarlyStopping(
    monitor='val_mae',
    patience=15,
    restore_best_weights=True,
    verbose=1
)

reduce_lr = callbacks.ReduceLROnPlateau(
    monitor='val_mae',
    factor=0.5,
    patience=7,
    min_lr=1e-6,
    verbose=1
)

# ===== HYPERPARAMETER TUNING =====
tuner = kt.BayesianOptimization(
    build_professional_model,
    objective='val_mae',
    max_trials=15,
    executions_per_trial=1,
    directory='final_tuning',
    project_name='professional_multi_output',
    overwrite=True
)

print("Starting Hyperparameter Search...")
print("=" * 50)

tuner.search(
    [X_seq_train, X_static_train], 
    y_train,
    epochs=100,
    validation_data=([X_seq_test, X_static_test], y_test),
    callbacks=[early_stopping, reduce_lr],
    verbose=1
)

# Get Best Model
best_model = tuner.get_best_models(num_models=1)[0]
best_hp = tuner.get_best_hyperparameters(num_trials=1)[0]

print("\n" + "=" * 50)
print("BEST HYPERPARAMETERS:")
print("=" * 50)
for param in ['lstm_units_1', 'lstm_units_2', 'dense_static', 'dense_combined', 
              'dropout', 'l2_reg', 'learning_rate', 'use_bidirectional']:
    print(f"  {param}: {best_hp.get(param)}")

best_model.summary()
```
